```python
import pandas as pd

df = pd.read_csv("../../data/developer_ai_learning_raw.csv")
df.shape
df.head
```

```python
df.duplicated().sum()
```

```python
df.dtypes
```

```python
df["YearsCode"] = pd.to_numeric(df["YearsCode"], errors="coerce").astype("Int64")
df.dtypes
```

```python
cols = ["LearnCode","AILearnHow"]

for col in cols:
    df[col] = df[col].str.split(";")
```

```python
df.to_json(
    "cleaned_data.json",
    orient="records",
    lines=True
)
```
